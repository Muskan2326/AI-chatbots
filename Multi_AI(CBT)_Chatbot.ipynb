```python
def mock_llm(prompt: str, context: str = "") -> str:
    """
    Simulates the behavior of a Large Language Model (LLM) for processing user input
    and generating agent-specific responses based on keywords.

    Args:
        prompt (str): The user's input or an agent's query.
        context (str, optional): Additional information from other agents. Defaults to "".

    Returns:
        str: A simulated LLM response.
    """
    text = (prompt + " " + context).lower()

    # Keywords for Emotional State
    emotional_keywords = {
        'sad': "I hear you're feeling sad. It's okay to feel that way. Can you tell me more about it?",
        'anxious': "Feeling anxious can be tough. What's on your mind right now?",
        'happy': "That's wonderful to hear you're feeling happy! What's making you feel good?",
        'stressed': "It sounds like you're under stress. Let's explore what might be contributing to that."
    }

    for keyword, response in emotional_keywords.items():
        if keyword in text:
            return response

    # Keywords for CBT Concepts
    cbt_keywords = {
        'thought pattern': "Let's try to identify any recurring thought patterns you might have.",
        'challenge thought': "Have you considered challenging that thought? What evidence supports it?",
        'behavior': "How do your behaviors connect to your feelings and thoughts?"
    }

    for keyword, response in cbt_keywords.items():
        if keyword in text:
            return response

    # Keywords for Psychodynamic Concepts
    psychodynamic_keywords = {
        'childhood': "Tell me about your childhood. How do you think past experiences might be influencing you now?",
        'past': "Sometimes understanding our past can shed light on our present feelings.",
        'relationship': "Let's explore your relationships and how they might be affecting you.",
        'unconscious': "It's interesting to consider what might be operating beneath the surface, perhaps unconsciously."
    }

    for keyword, response in psychodynamic_keywords.items():
        if keyword in text:
            return response

    # Generic conversational response if no specific keywords are matched
    return "That's an interesting point. Could you elaborate further?"

print("Mock LLM function 'mock_llm' has been defined.")
```

```python
import re

def analyze_emotional_state(user_input: str) -> dict:
    """
    Analyzes user input to identify emotional states and assigns an intensity score.

    Args:
        user_input (str): The user's text input.

    Returns:
        dict: A dictionary containing identified emotions, categories, and intensity scores.
              Returns a neutral state if no specific emotion is detected.
    """
    text = user_input.lower()
    emotional_states = {
        'sadness': {'keywords': ['sad', 'unhappy', 'depressed', 'down', 'mournful'], 'category': 'negative', 'intensity': 0.7},
        'joy': {'keywords': ['happy', 'joyful', 'excited', 'glad', 'upbeat'], 'category': 'positive', 'intensity': 0.8},
        'anger': {'keywords': ['angry', 'frustrated', 'irritated', 'mad', 'annoyed'], 'category': 'negative', 'intensity': 0.9},
        'anxiety': {'keywords': ['anxious', 'nervous', 'worried', 'stressed', 'frightened'], 'category': 'negative', 'intensity': 0.75},
        'surprise': {'keywords': ['surprised', 'shocked', 'astonished'], 'category': 'neutral', 'intensity': 0.6},
        'fear': {'keywords': ['scared', 'afraid', 'terrified', 'fearful'], 'category': 'negative', 'intensity': 0.85}
    }

    detected_emotions = []

    for emotion_name, data in emotional_states.items():
        for keyword in data['keywords']:
            if re.search(r'\b' + re.escape(keyword) + r'\b', text):
                # Mock intensity score (e.g., higher for more direct matches or negative emotions)
                # For simplicity, using the predefined intensity for now.
                detected_emotions.append({
                    'emotion': emotion_name,
                    'category': data['category'],
                    'intensity': data['intensity']
                })
                break # Move to the next emotion after first keyword match

    if not detected_emotions:
        return {
            'emotion': 'neutral',
            'category': 'neutral',
            'intensity': 0.2
        }

    # In a real scenario, you might want to aggregate or pick the most dominant emotion
    # For this mock, we'll return the first detected, or a list of all detected if needed
    return {"detected_emotions": detected_emotions}

print("Emotional State Agent function 'analyze_emotional_state' has been defined.")
```

```python
import re

def cbt_agent_process(user_input: str, emotional_state: dict = None) -> str:
    """
    Identifies common cognitive distortions in user input and provides CBT-based interventions.

    Args:
        user_input (str): The user's text input.
        emotional_state (dict, optional): Emotional state detected by the Emotional State Agent.

    Returns:
        str: A CBT-based intervention or reflection.
    """
    text = user_input.lower()
    response = ""

    # Define common cognitive distortions and associated keywords/responses
    cognitive_distortions = {
        'all-or-nothing thinking': {
            'keywords': ['always', 'never', 'everything', 'nothing', 'fail', 'perfect'],
            'response': "It sounds like you're seeing things in 'all-or-nothing' terms. Can we explore if there are any shades of gray or other possibilities?"
        },
        'catastrophizing': {
            'keywords': ['worst', 'disaster', 'terrible', 'ruined', 'awful', 'dreadful'],
            'response': "When you think about the worst possible outcome, it's natural to feel overwhelmed. What's the most realistic outcome, even if it's not ideal?"
        },
        'overgeneralization': {
            'keywords': ['every time', 'always happens', 'another failure', 'typical'],
            'response': "It seems like you're generalizing from one or two events. Can we consider if this truly happens every time, or if there might be exceptions?"
        },
        'mental filter': {
            'keywords': ['only see', "can't help but notice", 'focus on', 'ignore the good'],
            'response': "Are you perhaps focusing on only the negative aspects and filtering out the positive? Let's try to notice the whole picture."
        },
        'should statements': {
            'keywords': ['should', 'must', 'ought to'],
            'response': "You're using 'should' statements. Who set these rules, and what happens if they're not met?"
        },
        'personalization': {
            'keywords': ['my fault', 'because of me', 'i caused'],
            'response': "It sounds like you're taking full responsibility for something. Are there other factors or people involved that might share some of the responsibility?"
        },
        'mind reading': {
            'keywords': ['they think', 'know what they mean', "they're judging"],
            'response': "It seems you're assuming what others are thinking. Have you considered asking them directly, or are there other interpretations of their actions?"
        }
    }

    # Check for cognitive distortions
    for distortion_type, data in cognitive_distortions.items():
        for keyword in data['keywords']:
            if re.search(r'\b' + re.escape(keyword) + r'\b', text):
                response = data['response']
                # Optional: Enhance response based on emotional state if provided
                if emotional_state and 'sadness' in [e['emotion'] for e in emotional_state.get('detected_emotions', [])]:
                    response += " I understand this can be especially tough when you're feeling down."
                return response

    # If no specific distortion detected, provide a general CBT-related probing question
    if not response:
        if emotional_state and emotional_state.get('emotion') == 'neutral' and not emotional_state.get('detected_emotions'):
             return "Let's explore your thoughts. What thoughts are running through your mind right now?"
        else:
            # General CBT prompt, potentially influenced by detected emotions
            if emotional_state and emotional_state.get('detected_emotions'):
                main_emotion = emotional_state['detected_emotions'][0]['emotion'] # Take first detected emotion
                if main_emotion in ['sadness', 'anxiety']:
                    return f"It sounds like you're feeling {main_emotion}. What specific thoughts are connected to this feeling?"
            return "Can you tell me more about what you're thinking and how it's making you feel?"


print("CBT Agent function 'cbt_agent_process' has been defined.")
```

```python
import re

def psychodynamic_agent_process(user_input: str, emotional_state: dict = None) -> str:
    """
    Analyzes user input for psychodynamic themes and generates reflective responses.

    Args:
        user_input (str): The user's text input.
        emotional_state (dict, optional): Emotional state detected by the Emotional State Agent.

    Returns:
        str: A psychodynamic-based intervention or reflection.
    """
    text = user_input.lower()

    psychodynamic_themes = {
        'childhood': {
            'keywords': ['childhood', 'growing up', 'early years', 'parents', 'family dynamics'],
            'response': "It sounds like your childhood experiences are on your mind. How do you feel those early experiences might be connected to what you're feeling now?"
        },
        'past experiences': {
            'keywords': ['past', 'history', 'used to be', 'old self'],
            'response': "You're reflecting on your past experiences. How do you see those moments influencing your current thoughts or reactions?"
        },
        'relationships': {
            'keywords': ['relationships', 'partner', 'friend', 'mother', 'father', 'sibling', 'interact', 'connection'],
            'response': "Let's explore your relationships. How do your interactions with others, especially significant figures, play a role in your emotional landscape?"
        },
        'patterns': {
            'keywords': ['pattern', 'always happens', 'recurring', 'cycle', 'repeat'],
            'response': "You've noticed a pattern. What feelings or situations seem to repeat themselves, and what might be beneath the surface of these repetitions?"
        },
        'unconscious': {
            'keywords': ['unconscious', 'subconscious', 'dream', 'deep down', 'hidden'],
            'response': "Sometimes there are influences at play that we're not fully aware of. Have you considered what might be operating on an unconscious level?"
        }
    }

    for theme, data in psychodynamic_themes.items():
        for keyword in data['keywords']:
            if re.search(r'\b' + re.escape(keyword) + r'\b', text):
                response = data['response']
                # Optional: Add emotional context if available
                if emotional_state and emotional_state.get('detected_emotions'):
                    main_emotion = emotional_state['detected_emotions'][0]['emotion']
                    response += f" Especially when you're feeling {main_emotion}, these connections can be quite profound."
                return response

    # Default psychodynamic probing question if no specific theme is detected
    return "Sometimes understanding our past can shed light on our present. Would you like to explore how your history might be influencing you today?"

print("Psychodynamic Agent function 'psychodynamic_agent_process' has been defined.")
```

```python
import re

def safety_escalation_agent_process(user_input: str) -> dict:
    """
    Checks user input for keywords indicating crisis or safety concerns.

    Args:
        user_input (str): The user's text input.

    Returns:
        dict: A dictionary indicating if escalation is needed and a corresponding message.
    """
    text = user_input.lower()

    crisis_keywords = [
        'harm',
        'suicide',
        'kill myself',
        'danger',
        'emergency',
        'hurt myself',
        'ending it all',
        'take my own life',
        'die',
        'can\'t go on',
        'no reason to live'
    ]

    for keyword in crisis_keywords:
        if re.search(r'\b' + re.escape(keyword) + r'\b', text):
            return {
                'escalate': True,
                'message': "I detect words that suggest you might be in danger. Please reach out to a crisis hotline or emergency services immediately. You are not alone, and help is available. Here are some resources: [National Suicide Prevention Lifeline: 988, Crisis Text Line: Text HOME to 741741]."
            }

    return {
        'escalate': False,
        'message': 'No immediate safety concerns detected.'
    }

print("Safety & Escalation Agent function 'safety_escalation_agent_process' has been defined.")
```

```python
import re

def suggestion_agent_process(user_input: str, emotional_state: dict = None) -> str:
    """
    Provides basic suggestions based on detected emotional states or therapeutic themes.

    Args:
        user_input (str): The user's text input.
        emotional_state (dict, optional): Emotional state detected by the Emotional State Agent.

    Returns:
        str: A string containing the suggestion or a generic supportive statement.
    """
    text = user_input.lower()
    suggestions = {
        'stress': {
            'keywords': ['stressed', 'overwhelmed', 'anxious', 'pressure', 'burnout'],
            'suggestion': "It sounds like you're experiencing stress. Taking a few deep breaths can sometimes help. Would you like to try a simple breathing exercise?"
        },
        'sleep': {
            'keywords': ['sleep', 'insomnia', 'tired', 'can\'t sleep', 'bedtime'],
            'suggestion': "If you're having trouble with sleep, establishing a consistent bedtime routine can be very helpful. Would you like some tips on creating one?"
        },
        'anger': {
            'keywords': ['angry', 'frustrated', 'mad', 'rage', 'irritated'],
            'suggestion': "When you feel anger building, it can be useful to pause and identify the triggers. Perhaps we can explore some healthy ways to express these feelings."
        },
        'sadness': {
            'keywords': ['sad', 'depressed', 'unhappy', 'down', 'grief'],
            'suggestion': "It's understandable to feel sad. Sometimes connecting with a supportive friend or engaging in a comforting activity can help. What brings you a sense of comfort?"
        },
        'focus': {
            'keywords': ['focus', 'distracted', 'concentrate', 'attention'],
            'suggestion': "If you're struggling with focus, breaking tasks into smaller steps or using a \"pomodoro technique\" might be beneficial. Would you like to know more?"
        }
    }

    # Check user input for themes
    for theme, data in suggestions.items():
        for keyword in data['keywords']:
            if re.search(r'\b' + re.escape(keyword) + r'\b', text):
                return data['suggestion']

    # Check emotional state if provided
    if emotional_state and emotional_state.get('detected_emotions'):
        for emotion_data in emotional_state['detected_emotions']:
            emotion = emotion_data['emotion']
            if emotion == 'anxiety':
                return "Given that you're feeling anxious, perhaps practicing mindfulness or progressive muscle relaxation could offer some relief?"
            elif emotion == 'sadness':
                return "It sounds like you're experiencing sadness. Sometimes, talking about your feelings or engaging in a pleasant activity can be helpful. What do you think?"
            elif emotion == 'joy':
                return "It's wonderful to hear you're feeling joyful! How can we build on this positive emotion or integrate more of what makes you happy into your day?"

    # Generic supportive statement if no specific suggestion is generated
    return "I'm here to listen. What's on your mind today, and how can I support you?"

print("Suggestion Agent function 'suggestion_agent_process' has been defined.")
```

```python
import collections

# Using a deque for a fixed-size history
interaction_history = collections.deque(maxlen=10) # Stores last 10 interactions

def progress_agent_process(user_input: str, emotional_state: dict, cbt_insight: str, psychodynamic_insight: str) -> str:
    """
    Simulates tracking user interactions and identified themes, returning a summary.

    Args:
        user_input (str): The user's current text input.
        emotional_state (dict): Emotional state detected by the Emotional State Agent.
        cbt_insight (str): Insight/response from the CBT Agent.
        psychodynamic_insight (str): Insight/response from the Psychodynamic Agent.

    Returns:
        str: A summary of recent interactions, themes, or emotional states.
    """
    global interaction_history

    # Append current interaction to history
    current_interaction = {
        'user_input': user_input,
        'emotional_state': emotional_state,
        'cbt_insight': cbt_insight,
        'psychodynamic_insight': psychodynamic_insight
    }
    interaction_history.append(current_interaction)

    # Summarize recent interactions
    if not interaction_history:
        return "No interactions yet. Let's start building your progress history."

    summary_parts = []
    summary_parts.append("--- Progress Agent Summary ---")
    summary_parts.append(f"Total interactions tracked: {len(interaction_history)}")

    # Summarize emotional states from recent interactions
    all_detected_emotions = []
    for entry in interaction_history:
        if entry['emotional_state'] and entry['emotional_state'].get('detected_emotions'):
            for emotion_data in entry['emotional_state']['detected_emotions']:
                all_detected_emotions.append(emotion_data['emotion'])

    if all_detected_emotions:
        emotion_counts = collections.Counter(all_detected_emotions)
        most_common_emotions = emotion_counts.most_common(3)
        summary_parts.append(f"Most common recent emotions: {', '.join([f'{e[0]} ({e[1]} times)' for e in most_common_emotions])}")
    else:
        summary_parts.append("Recent emotional states: N/A")

    # Summarize last few user inputs
    last_n_inputs = list(interaction_history)[-3:] # Get last 3, or fewer if history is shorter
    if last_n_inputs:
        summary_parts.append("\nRecent user concerns:")
        for i, entry in enumerate(last_n_inputs):
            summary_parts.append(f"  {i+1}. '{entry['user_input']}'")

    return "\n".join(summary_parts)

print("Progress Agent function 'progress_agent_process' has been defined and initialized with a history tracker.")
```

```python
class IntegrationAgent:
    """
    Orchestrates the flow between different therapeutic AI agents.
    Receives user input, dispatches it to specialized agents, and combines their outputs
    into a cohesive context for the Mock LLM.
    """
    def __init__(self, mock_llm_func, emotional_state_func, cbt_func, psychodynamic_func,
                 safety_escalation_func, suggestion_func, progress_func):
        self.mock_llm = mock_llm_func
        self.analyze_emotional_state = emotional_state_func
        self.cbt_agent_process = cbt_func
        self.psychodynamic_agent_process = psychodynamic_func
        self.safety_escalation_agent_process = safety_escalation_func
        self.suggestion_agent_process = suggestion_func
        self.progress_agent_process = progress_func

    def process_user_input(self, user_input: str) -> dict:
        """
        Processes user input by orchestrating calls to various agents and integrating their responses.

        Args:
            user_input (str): The user's text input.

        Returns:
            dict: An integrated context containing outputs from all agents, ready for the Mock LLM,
                  or a safety escalation message.
        """
        # 1. Safety & Escalation Check (immediate return if escalation is needed)
        safety_response = self.safety_escalation_agent_process(user_input)
        if safety_response['escalate']:
            return {
                'status': 'escalated',
                'response': safety_response['message']
            }

        # 2. Emotional State Analysis
        emotional_state = self.analyze_emotional_state(user_input)

        # 3. CBT Agent Processing
        cbt_insight = self.cbt_agent_process(user_input, emotional_state)

        # 4. Psychodynamic Agent Processing
        psychodynamic_insight = self.psychodynamic_agent_process(user_input, emotional_state)

        # 5. Suggestion Agent Processing
        suggestion_response = self.suggestion_agent_process(user_input, emotional_state)

        # 6. Progress Agent Processing (requires outputs from other agents)
        progress_summary = self.progress_agent_process(user_input, emotional_state, cbt_insight, psychodynamic_insight)

        # 7. Combine all outputs into a cohesive context for the Mock LLM
        # This context can be a string or a structured dictionary
        integrated_context = {
            'status': 'processed',
            'user_input': user_input,
            'emotional_state': emotional_state,
            'cbt_insight': cbt_insight,
            'psychodynamic_insight': psychodynamic_insight,
            'suggestion': suggestion_response,
            'progress_summary': progress_summary,
            'mock_llm_context': (
                f"User said: '{user_input}'. "
                f"Emotional state detected: {emotional_state}. "
                f"CBT insight: {cbt_insight}. "
                f"Psychodynamic insight: {psychodynamic_insight}. "
                f"Suggestion: {suggestion_response}. "
                f"Progress summary: {progress_summary}"
            )
        }

        return integrated_context

print("IntegrationAgent class has been defined.")
```

```python
class CompanionAgent:
    """
    Acts as the conversational interface, utilizing the IntegrationAgent to process
    user input and the mock_llm to generate the final, natural language response.
    """
    def __init__(self, integration_agent_instance: 'IntegrationAgent', mock_llm_func):
        self.integration_agent = integration_agent_instance
        self.mock_llm = mock_llm_func

    def get_response(self, user_input: str) -> str:
        """
        Processes user input through the IntegrationAgent and generates a final response
        using the mock_llm.

        Args:
            user_input (str): The user's text input.

        Returns:
            str: The final natural language response to the user, or a safety escalation message.
        """
        # Process user input through the IntegrationAgent
        integrated_context = self.integration_agent.process_user_input(user_input)

        # Check for safety escalation
        if integrated_context['status'] == 'escalated':
            return integrated_context['response']

        # If not escalated, generate response using Mock LLM
        if integrated_context['status'] == 'processed':
            mock_llm_input_context = integrated_context['mock_llm_context']
            # For a more empathetic tone, we can prepend a general empathetic phrase
            # to the LLM's prompt if the emotional state is negative.
            empathy_prefix = ""
            if integrated_context['emotional_state'] and integrated_context['emotional_state'].get('detected_emotions'):
                for emotion_data in integrated_context['emotional_state']['detected_emotions']:
                    if emotion_data['category'] == 'negative' and emotion_data['intensity'] > 0.5:
                        empathy_prefix = "I understand this is a difficult situation. "
                        break

            final_response = self.mock_llm(prompt=user_input, context=empathy_prefix + mock_llm_input_context)
            return final_response

        return "I'm sorry, I couldn't process that request." # Fallback

print("CompanionAgent class has been defined.")
```

```python
print("Initializing all agents for demonstration...")

# Initialize the IntegrationAgent with all necessary mock functions
integration_agent_instance = IntegrationAgent(
    mock_llm_func=mock_llm,
    emotional_state_func=analyze_emotional_state,
    cbt_func=cbt_agent_process,
    psychodynamic_func=psychodynamic_agent_process,
    safety_escalation_func=safety_escalation_agent_process,
    suggestion_func=suggestion_agent_process,
    progress_func=progress_agent_process
)

# Initialize the CompanionAgent with the IntegrationAgent instance and mock_llm
companion_agent = CompanionAgent(integration_agent_instance, mock_llm)

def simulate_conversation(interactions):
    print("\n--- Starting Simulated Conversation ---")
    for i, user_input in enumerate(interactions):
        print(f"\nUser [{i+1}]: {user_input}")
        response = companion_agent.get_response(user_input)
        print(f"AI [{i+1}]: {response}")
    print("\n--- End Simulated Conversation ---")


# Example conversation interactions
conversation_interactions = [
    "I'm feeling really sad today, everything seems so difficult and I just want to stay in bed.",
    "I always mess things up. I'm such a failure.",
    "My parents always made me feel like I wasn't good enough when I was a child.",
    "I'm so stressed about work, I can't sleep at night.",
    "I feel like there's no point in going on, I want to end it all.",
    "Just feeling a bit down, but trying to cope."
]

simulate_conversation(conversation_interactions)
```

```python
import re

def safety_escalation_agent_process(user_input: str) -> dict:
    """
    Checks user input for keywords indicating crisis or safety concerns.

    Args:
        user_input (str): The user's text input.

    Returns:
        dict: A dictionary indicating if escalation is needed and a corresponding message.
    """
    text = user_input.lower()

    # Keywords for single words, requiring explicit word boundaries
    crisis_keywords_single = [
        'harm',
        'suicide',
        'danger',
        'emergency',
        'die'
    ]

    # Keywords for phrases, where direct substring search or slightly more flexible regex is better
    crisis_keywords_phrases = [
        'kill myself',
        'hurt myself',
        'ending it all',
        'end it all',
        'take my own life',
        'can\'t go on',
        'no point in going on',
        'no reason to live',
        'give up'
    ]

    # Check for phrases first
    for keyword in crisis_keywords_phrases:
        if re.search(re.escape(keyword), text): # Use re.escape for special chars, but no  for flexible phrase matching
            return {
                'escalate': True,
                'message': "I detect words that suggest you might be in danger. Please reach out to a crisis hotline or emergency services immediately. You are not alone, and help is available. Here are some resources: [National Suicide Prevention Lifeline: 988, Crisis Text Line: Text HOME to 741741]."
            }

    # Check for single words with strict word boundaries
    for keyword in crisis_keywords_single:
        if re.search(r'\b' + re.escape(keyword) + r'\b', text): # Use word boundaries for single words
            return {
                'escalate': True,
                'message': "I detect words that suggest you might be in danger. Please reach out to a crisis hotline or emergency services immediately. You are not alone, and help is available. Here are some resources: [National Suicide Prevention Lifeline: 988, Crisis Text Line: Text HOME to 741741]."
            }

    return {
        'escalate': False,
        'message': 'No immediate safety concerns detected.'
    }

print("Safety & Escalation Agent function 'safety_escalation_agent_process' has been redefined with improved keyword detection.")
```

```python
import re

def mock_llm(prompt: str, context: str = "") -> str:
    """
    Simulates the behavior of a Large Language Model (LLM) for processing user input
    and generating agent-specific responses based on insights from other agents.

    Args:
        prompt (str): The user's input.
        context (str, optional): Additional structured information from other agents.

    Returns:
        str: A simulated LLM response synthesizing agent outputs.
    """

    # Attempt to parse specific insights from the structured context string
    cbt_insight = ""
    psychodynamic_insight = ""
    suggestion = ""
    emotional_state_str = ""

    # Regular expressions to extract insights from the context string
    cbt_match = re.search(r"CBT insight: (.*?)(?=\. Psychodynamic insight:|$)", context)
    if cbt_match and cbt_match.group(1).strip() != "Can you tell me more about what you're thinking and how it's making you feel?":
        cbt_insight = cbt_match.group(1).strip()

    psycho_match = re.search(r"Psychodynamic insight: (.*?)(?=\. Suggestion:|$)", context)
    if psycho_match and psycho_match.group(1).strip() != "Sometimes understanding our past can shed light on our present. Would you like to explore how your history might be influencing you today?":
        psychodynamic_insight = psycho_match.group(1).strip()

    suggestion_match = re.search(r"Suggestion: (.*?)(?=\. Progress summary:|$)", context)
    if suggestion_match and suggestion_match.group(1).strip() != "I'm here to listen. What's on your mind today, and how can I support you?":
        suggestion = suggestion_match.group(1).strip()

    emotional_state_match = re.search(r"Emotional state detected: (.*?)(?=\. CBT insight:|$)", context)
    if emotional_state_match:
        emotional_state_str = emotional_state_match.group(1).strip().lower()

    # Prioritize responses based on agent insights
    if cbt_insight and not cbt_insight.startswith("Can you tell me more"): # Avoid generic CBT fallback
        return f"Regarding your thoughts, {cbt_insight}"

    if psychodynamic_insight and not psychodynamic_insight.startswith("Sometimes understanding"): # Avoid generic Psychodynamic fallback
        return f"Delving deeper, {psychodynamic_insight}"

    if suggestion and not suggestion.startswith("I'm here to listen"): # Avoid generic suggestion fallback
        return f"Here's a thought for you: {suggestion}"

    # Fallback to direct keyword matching on the original prompt for emotional states if no specific insights
    text = prompt.lower() # Use original prompt for direct emotional keywords

    emotional_keywords = {
        'sad': "I hear you're feeling sad. It's okay to feel that way. Can you tell me more about it?",
        'anxious': "Feeling anxious can be tough. What's on your mind right now?",
        'stressed': "It sounds like you're under stress. Let's explore what might be contributing to that.",
        'happy': "That's wonderful to hear you're feeling happy! What's making you feel good?"
    }

    for keyword, response in emotional_keywords.items():
        if keyword in text:
            return response

    # Generic conversational response if no specific keywords or agent insights are matched
    return "That's an interesting point. Could you elaborate further?"

print("Mock LLM function 'mock_llm' has been redefined to integrate agent insights.")
```

```python
import re

def mock_llm(prompt: str, context: str = "") -> str:
    """
    Simulates the behavior of a Large Language Model (LLM) for processing user input
    and generating agent-specific responses based on insights from other agents.

    Args:
        prompt (str): The user's input.
        context (str, optional): Additional structured information from other agents.

    Returns:
        str: A simulated LLM response synthesizing agent outputs.
    """

    # Attempt to parse specific insights from the structured context string
    cbt_insight = ""
    psychodynamic_insight = ""
    suggestion = ""
    emotional_state_str = ""

    cbt_match = re.search(r"CBT insight: (.*?)(?=\. Psychodynamic insight:|$)", context)
    if cbt_match and cbt_match.group(1).strip() != "Can you tell me more about what you're thinking and how it's making you feel?":
        cbt_insight = cbt_match.group(1).strip()

    psycho_match = re.search(r"Psychodynamic insight: (.*?)(?=\. Suggestion:|$)", context)
    if psycho_match and psycho_match.group(1).strip() != "Sometimes understanding our past can shed light on our present. Would you like to explore how your history might be influencing you today?":
        psychodynamic_insight = psycho_match.group(1).strip()

    suggestion_match = re.search(r"Suggestion: (.*?)(?=\. Progress summary:|$)", context)
    if suggestion_match and suggestion_match.group(1).strip() != "I'm here to listen. What's on your mind today, and how can I support you?":
        suggestion = suggestion_match.group(1).strip()

    emotional_state_match = re.search(r"Emotional state detected: (.*?)(?=\. CBT insight:|$)", context)
    if emotional_state_match:
        emotional_state_str = emotional_state_match.group(1).strip().lower()

    # Construct response by combining insights, prioritizing based on content relevance
    response_parts = []

    # Prepend empathetic prefix from CompanionAgent if present
    empathy_prefix_match = re.match(r"(I understand this is a difficult situation\. )", context)
    if empathy_prefix_match:
        response_parts.append(empathy_prefix_match.group(1))

    if psychodynamic_insight:
        response_parts.append(f"Delving deeper, {psychodynamic_insight}")

    if cbt_insight:
        response_parts.append(f"Regarding your thoughts, {cbt_insight}")

    if suggestion:
        # Integrate suggestion more smoothly based on existing insights or as a standalone
        if not response_parts:
            response_parts.append(f"Here's a thought for you: {suggestion}")
        else:
            response_parts.append(f"Also, consider this: {suggestion}")

    # Fallback to direct keyword matching on the original prompt for emotional states
    # if no specific agent insights were integrated to prevent generic fallback too soon.
    if not response_parts or len(response_parts) == 1 and "I understand this is a difficult situation" in response_parts[0]:
        text = prompt.lower()
        emotional_keywords = {
            'sad': "I hear you're feeling sad. It's okay to feel that way. Can you tell me more about it?",
            'anxious': "Feeling anxious can be tough. What's on your mind right now?",
            'stressed': "It sounds like you're under stress. Let's explore what might be contributing to that.",
            'happy': "That's wonderful to hear you're feeling happy! What's making you feel good?"
        }

        for keyword, response in emotional_keywords.items():
            if keyword in text:
                if not response_parts:
                    response_parts.append(response)
                else:
                    # Append emotional response if it complements existing parts
                    response_parts.append(f" Also, {response[0].lower() + response[1:]}") # make first letter lowercase for smoother join
                break

    # If still no specific response, provide a generic conversational fallback
    if not response_parts:
        return "That's an interesting point. Could you elaborate further?"

    return " ".join(response_parts).strip()

print("Mock LLM function 'mock_llm' has been further refined for integrated responses.")
```

## Summary:

### Data Analysis Key Findings

*   **Safety Escalation Agent:** A `safety_escalation_agent_process` function was implemented to detect crisis keywords (e.g., 'harm', 'suicide', 'danger', 'no point in going on') in user input. It returns an escalation flag (`'escalate': True`) and a safety message with crisis resources if such keywords are found, otherwise, it indicates 'No immediate safety concerns detected.' The function was refined to accurately detect multi-word phrases by adjusting regex patterns.
*   **Suggestion Agent:** A `suggestion_agent_process` function was developed to provide basic therapeutic suggestions. It uses keyword matching on user input for themes like 'stress' or 'sleep', and also considers detected `emotional_state` (e.g., 'anxiety', 'sadness') to offer relevant advice like "Taking a few deep breaths can sometimes help" for stress.
*   **Progress Agent:** A `progress_agent_process` function was created to simulate tracking user interactions. It maintains a history (a `deque` with a `maxlen` of 10) of `user_input`, `emotional_state`, and insights from CBT and Psychodynamic agents. This agent summarizes recent emotional states (e.g., "Most common recent emotions: sad (1 time)") and the last few user inputs.
*   **Integration Agent:** An `IntegrationAgent` class was implemented to orchestrate the workflow. It receives user input, first dispatches it to the `safety_escalation_agent_process` (returning an immediate escalation if triggered), then to the Emotional State, CBT, Psychodynamic, and Suggestion agents. Finally, it uses the outputs of these agents along with the current user input to update the `progress_agent_process` and compile a comprehensive `integrated_context` for the Mock LLM.
*   **Companion Agent:** The `CompanionAgent` class acts as the conversational interface. It uses the `IntegrationAgent` to process user input and then leverages a `mock_llm` function to generate the final natural language response. It includes logic to prepend an empathetic phrase if negative emotions are detected.
*   **Mock LLM Refinement:** The `mock_llm` function was iteratively refined to parse and prioritize insights from the CBT, Psychodynamic, and Suggestion agents within the `integrated_context`. This enabled the LLM to generate more nuanced and combined responses, moving beyond generic replies to specifically address user concerns with relevant therapeutic insights, and also handling empathetic phrasing.
*   **Simulated Conversation:** An interactive chat loop was successfully implemented, demonstrating the system's ability to process user inputs, trigger safety escalations for phrases like "I feel like there's no point in going on, I want to end it all," and provide context-aware, integrated responses from the various agents through the `CompanionAgent`.

### Insights or Next Steps

*   The modular agent-based architecture significantly enhances the system's maintainability and extensibility, allowing individual therapeutic components to be developed and improved independently.
*   Further development of the `mock_llm` to incorporate more sophisticated natural language generation techniques could lead to more fluid and human-like conversational responses, moving beyond mere concatenation of agent insights.
